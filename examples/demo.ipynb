# Two-Body Simulation - Demo
Notebook de demostración de la simulación gravitacional de dos cuerpos.

## 1. Imports y parámetros iniciales

```
import sys
sys.path.append('..')
import numpy as np
import matplotlib.pyplot as plt
from two_body import compute_acceleration, compute_energy
from integrators import euler_step, verlet_step, rk4_step
from plots import plot_trajectory, plot_energy

m1, m2 = 1.0, 1.0
r1 = np.array([1.0, 0.0])
r2 = np.array([-1.0, 0.0])
v1 = np.array([0.0, 0.5])
v2 = np.array([0.0, -0.5])
dt = 0.01
N = 1000
integrador = verlet_step
```

## 2. Bucle de simulación

```
r1_list = [r1.copy()]
r2_list = [r2.copy()]
E_list  = [compute_energy(r1, r2, v1, v2, m1, m2)]

for i in range(N):
    r1, r2, v1, v2 = integrador(r1, r2, v1, v2, m1, m2, dt)
    r1_list.append(r1.copy())
    r2_list.append(r2.copy())
    E_list.append(compute_energy(r1, r2, v1, v2, m1, m2))

r1_list = np.array(r1_list)
r2_list = np.array(r2_list)
print(f'Simulación completada! Integrador: {integrador.__name__}')
```

## 3. Trayectorias

```
plot_trajectory(r1_list, r2_list, title=f'Trayectoria - {integrador.__name__}')
plt.show()
```

## 4. Conservación de energía

```
plot_energy(E_list, title=f'Energía - {integrador.__name__}')
plt.show()
```

## 5. Comparación de integradores

```
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
integradores = [euler_step, verlet_step, rk4_step]

for ax, integ in zip(axes, integradores):
    r1 = np.array([1.0, 0.0])
    r2 = np.array([-1.0, 0.0])
    v1 = np.array([0.0, 0.5])
    v2 = np.array([0.0, -0.5])
    rl1, rl2 = [r1.copy()], [r2.copy()]

    for i in range(N):
        r1, r2, v1, v2 = integ(r1, r2, v1, v2, m1, m2, dt)
        rl1.append(r1.copy())
        rl2.append(r2.copy())

    rl1, rl2 = np.array(rl1), np.array(rl2)
    ax.plot(rl1[:, 0], rl1[:, 1], label='Cuerpo 1')
    ax.plot(rl2[:, 0], rl2[:, 1], label='Cuerpo 2')
    ax.set_title(integ.__name__)
    ax.set_aspect('equal')
    ax.grid(True)
    ax.legend()

plt.suptitle('Comparación de integradores')
plt.tight_layout()
plt.show()
```
